**Q.2  python program using sample.txt**


```python
def exception_handling():
  try:
    with open("sample.txt" , "r") as f:
      content = f.read()
  except FileNotFoundError:
    print("File not found")
```

The above where the exception handling is done , here i have considered the error FileNotFoundError which mostly occur when text file is not uploaded aor found during the read file operation.
If so happens i have created a texy display to alert the user for file not found message.

**reading the file sample.txt**

```python
with open("sample.txt" , "r") as f:
  content = f.read()
  print(content)
```

```python
# convert str into list of words
words = content.split()
words
```

**calculating the word frequency within the text**

```python
word_freq={}
c=0
for word in words: # loop over each element in list
  if word in word_freq: # if word present in list then
    word_freq[word] += 1  #  increment the count where the word is already detected
  else:
    word_freq[word]=1  # if word not in list then add the word with count 1
```

```python
print(word_freq)
```

here the word frequencies are:
*  **Python** appears 3 times
*  **is** appears 1 times
*  **simple** appears 1 times
*  **powerful** appears 1 times
*  **fun** appears 1 times

```python
with open("output.txt","w") as f:
  for word , freq in word_freq.items():
    f.write(f"{word} : {freq}\n")
```

exported the word frequency into the txt file named output where it is represented as:
*  word : frequency count
